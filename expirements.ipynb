```python
import os
import pandas as pd
from mutagen.mp3 import MP3

path = "data"
audio_data = []

for root, dirs, files in os.walk(path):
    folder_date = os.path.basename(root)  # ✅ folder name as date

    for file in files:
        if file.lower().endswith(".mp3"):
            file_path = os.path.join(root, file)

            try:
                audio = MP3(file_path)
                duration_seconds = int(audio.info.length)  # in seconds

                audio_data.append({
                    "Date": folder_date,
                    "File Name": file,
                    "Duration (Seconds)": duration_seconds
                })

            except Exception as e:
                print(f"Error processing {file}: {e}")

df = pd.DataFrame(audio_data)
df.to_excel("audio_report.xlsx", index=False)
```
